# Break.
- used to terminate the loop when encountered.

# continue.
- execution in the current iteration(terms)& continue execution of loop with the next iteration.

```python
#break.

i = 1
while i <= 5:
    print(i)
    if(i == 3):
        break
    i +=1

print("End of Loop.")    
```

```python
#continue.

i = 0
while i <= 5:
    if(i == 3):
        i += 1
        continue#skip the value.
    print(i)
    i += 1
```

# For Loop.
- loops are used for sequential travesal for traversing list, string; tuples etc.

```python
#for loop.

list = [1,2,3]
for el in list:
    print(el)
```

```python
# why we use else for loop.

str = "Apna college"

for char in str:
    if (char == 'o'):
        print("o found")
        break
    print(char)
else:# if we dont give else and without else there will be in the end always End. but with else there is only end when there is condtion.
    print("End")   
```

# Range()
- Range functions return a sequence of numbers starting from 0 by default and increment by 1.(by default,and stops before a specified numbers)

- range(start?,stop,step?)

```python
#range(stop)
for el in range(10):#range stop. not included 10.
      print(el)
```

```python
#range(start,stop)
for el in range(2,10):
    print(el)
```

```python
# for el in range (2,10,2):#range (start,stop,step) kitne se increment krna. it negative also.

for el in range (2,10,2):
    print(el)
```
