# __init__ function():-
 CONSTRUCTOR (__init__)
✅ What is a Constructor?

- 👉 A constructor is a special function that runs automatically when object is created.

- In Python, its name is always
- __init__()
- ✅ Why do we need __init__?

- To:
✅ Store data inside the object
✅ Initialize values

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Shreyaa", 18)

print(s1.name)
print(s1.age)

```

# 📘 CHAPTER 8: THE 4 PILLARS OF OOPS (VERY IMPORTANT)

- These are asked in every interview.

# ✅ 1️⃣ ENCAPSULATION (Binding + Protection)
-✅ Meaning:

- 👉 Keeping data and methods together in one class and protecting data from outside misuse.

- Example:

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def show_balance(self):
        print(self.balance)

b = Bank(5000)
b.show_balance()
```

# ✅ Access Modifiers in Encapsulation
| Type      | Syntax | Meaning                |
| --------- | ------ | ---------------------- |
| Public    | name   | Accessible everywhere  |
| Protected | _name  | Inside class & child   |
| Private   | __name | Only inside same class |


```python
#Private Variable Example
class Bank:
    def __init__(self):
        self.__balance = 10000   # private

    def show(self):
        print(self.__balance)

b = Bank()
b.show()

#This will give error:
#print(b.__balance)
```

# 2️⃣ INHERITANCE (Parent → Child)
# ✅ Meaning:

- 👉 One class takes properties of another class.

```python
class Parent:
    def house(self):
        print("Parent's house")

class Child(Parent):
    def bike(self):
        print("Child's bike")

c = Child()
c.house()
c.bike()
```

# ✅ Types of Inheritance:

- (1)Single

- (2)Multiple

- (3)Multilevel

- (4)Hierarchical

```python
# multi level example
class A:
    def showA(self):
        print("A")

class B(A):
    def showB(self):
        print("B")

class C(B):
    def showC(self):
        print("C")

c = C()
c.showA()
c.showB()
c.showC()
```

# ✅ 3️⃣ POLYMORPHISM (Same Name, Different Work)
- ✅ Meaning:

- 👉 One function name, many forms

```python
#polymorphism.
class Bird:
    def sound(self):
        print("Bird makes sound")

class Dog(Bird):
    def sound(self):
        print("Dog barks")

b = Bird()
d = Dog()

b.sound()
d.sound()
```

✅ 4️⃣ ABSTRACTION (Hiding Internal Logic)
✅ Meaning:

- 👉 Show what is necessary, hide the internal implementation.

- Real Life Example:
- ATM Machine
- You only see:
- Insert card
- Enter PIN
- Withdraw money
- You don’t see banking servers
- In Python, abstraction is done using:
- abc module
- @abstractmethod

# 📘 CHAPTER 9: SUPER KEYWORD
- ✅ Use:
- 👉 Used to call parent class constructor or method

```python
class A:
    def __init__(self):
        print("Parent Constructor")

class B(A):
    def __init__(self):
        super().__init__()
        print("Child Constructor")

b = B()
```

# 📘 CHAPTER 10: COMPLETE REAL-LIFE PROGRAM

```python
class Employee:
    company = "Google"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Company:", Employee.company)

e1 = Employee("Amit", 50000)
e1.show()
```
